```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load data
train = pd.read_csv('./datathon-2026-round-1/sales.csv')
test = pd.read_csv('./datathon-2026-round-1/sample_submission.csv')
```

```python
train
```

```python
X = train.drop(['Date', 'Revenue'], axis=1)
y = train['Revenue']
# The test set should use the same feature columns as training data.
# Drop Date and Revenue so only numeric features remain.
X_test = test.drop(['Date', 'Revenue'], axis=1)

X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.2, random_state=88
)
```

```python
# Setting parameters can help improve model performance.
model = xgb.XGBRegressor(
    n_estimators=500,
    learning_rate=0.01,
    max_depth=8,
    min_child_weight=6,
    subsample=0.8,
    colsample_bytree=0.8,
    gamma=0.1,
    reg_alpha=0.05,
    reg_lambda=1,
    random_state=42,
    n_jobs=-1,
    # Remove the following parameters as they are for classification, not regression:
    # objective='multi:softprob',
    # num_class=3,
    # eval_metric='mlogloss'
)
model.fit(X_train, y_train,
    eval_set=[(X_train, y_train), (X_val, y_val)],
    verbose=50)
```

```python
# Evaluate on validation set
y_val_pred = model.predict(X_val)
val_rmse = np.sqrt(mean_squared_error(y_val, y_val_pred))
val_r2 = r2_score(y_val, y_val_pred)
print(f"Validation RMSE: {val_rmse:.2f}")
print(f"Validation R^2: {val_r2:.4f}")

# Evaluate on training set
y_train_pred = model.predict(X_train)
train_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))
train_r2 = r2_score(y_train, y_train_pred)
print(f"Training RMSE: {train_rmse:.2f}")
print(f"Training R^2: {train_r2:.4f}")
```

```python
y_test_pred = model.predict(X_test)
print(len(y_test_pred))
```

# Random Forest

```python
# Train Random Forest Regressor
rf_model = RandomForestRegressor(
    n_estimators=500,
    max_depth=10,
    min_samples_split=6,
    min_samples_leaf=4,
    max_features='sqrt',
    bootstrap=True,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

rf_model.fit(X_train, y_train)
print("\nRandom Forest training completed!")
```

```python
from sklearn.metrics import mean_squared_error, r2_score

# Evaluate Random Forest model on validation set
rf_val_pred = rf_model.predict(X_val)
rf_val_rmse = np.sqrt(mean_squared_error(y_val, rf_val_pred))
rf_val_r2 = r2_score(y_val, rf_val_pred)
print(f"Random Forest Validation RMSE: {rf_val_rmse:.2f}")
print(f"Random Forest Validation R^2: {rf_val_r2:.4f}")

# Evaluate on training set
rf_train_pred = rf_model.predict(X_train)
rf_train_rmse = np.sqrt(mean_squared_error(y_train, rf_train_pred))
rf_train_r2 = r2_score(y_train, rf_train_pred)
print(f"Random Forest Training RMSE: {rf_train_rmse:.2f}")
print(f"Random Forest Training R^2: {rf_train_r2:.4f}")
```

```python
# Random Forest predictions on test set
rf_test_pred = rf_model.predict(X_test)
```

```python
# Ensemble predictions - weighted average of XGBoost and Random Forest
xgb_weight = 0.5
rf_weight = 0.5

# Ensemble validation predictions
ensemble_val_pred = xgb_weight * y_val_pred + rf_weight * rf_val_pred

# Evaluate ensemble
ensemble_val_rmse = np.sqrt(mean_squared_error(y_val, ensemble_val_pred))
ensemble_val_r2 = r2_score(y_val, ensemble_val_pred)
print(f"Ensemble Validation RMSE: {ensemble_val_rmse:.2f}")
print(f"Ensemble Validation R^2: {ensemble_val_r2:.4f}")
```
